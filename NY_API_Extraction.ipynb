```python
NYT_API = ""
```

```python
import requests
import pandas as pd
import time

api_key = NYT_API  # your NYT API key
base_url = "https://api.nytimes.com/svc/search/v2/articlesearch.json"
movies = ["Atypical", "The Good Doctor", "Temple Grandin"]

all_articles = []

for movie in movies:
    page = 0
    while page < 3:  # up to 3 pages per movie
        params = {"q": f'"{movie}"', "api-key": api_key, "page": page}
        try:
            resp = requests.get(base_url, params=params)
            if resp.status_code == 429:
                print("⚠️ Rate limit reached. Sleeping 60s...")
                time.sleep(60)
                continue
            elif resp.status_code != 200:
                print(f"❌ API error {resp.status_code}. Skipping {movie}.")
                break

            docs = resp.json().get("response", {}).get("docs", [])
            if not docs:
                break

            for d in docs:
                all_articles.append({
                    "movie_title": movie,
                    "headline": d.get("headline", {}).get("main", ""),
                    "snippet": d.get("snippet", ""),
                    "url": d.get("web_url", ""),
                    "pub_date": d.get("pub_date", "")
                })

            page += 1
            time.sleep(0.5)  # avoid hitting rate limits

        except Exception as e:
            print(f"⚠️ Error fetching {movie}: {e}")
            break

df = pd.DataFrame(all_articles)
df.to_csv("nyt_movie_articles.csv", index=False)
print("Done! Articles saved to nyt_movie_articles.csv")


```
